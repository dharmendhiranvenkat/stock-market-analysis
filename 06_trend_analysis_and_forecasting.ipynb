```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\data\clean_analysis_utf8.csv"
)

# Show first rows
df.head()
```

```python
# Group company average sales
company_sales = df.groupby('company')['sales_value'].mean()

trend_result = []

for company in company_sales.index:
    
    sales = df[df['company'] == company]['sales_value'].dropna().values
    
    # Need at least 2 values
    if len(sales) >= 2:
        
        # Linear trend line
        slope = np.polyfit(range(len(sales)), sales, 1)[0]
        
        if slope > 0:
            trend = "UP 📈"
        elif slope < 0:
            trend = "DOWN 📉"
        else:
            trend = "FLAT ➖"
    
    else:
        trend = "Not Enough Data"
    
    trend_result.append([company, trend])

# Convert to table
trend_df = pd.DataFrame(
    trend_result,
    columns=['company', 'trend']
)

trend_df
```

```python
# Save trend results

trend_df.to_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\output\trend_analysis.csv",
    index=False
)

print("Trend analysis saved successfully!")
```
