```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\data\clean_analysis_utf8.csv"
)

# Show first rows
df.head()
```

```python
# Forecast next year revenue

forecast_result = []

for company in df['company'].unique():
    
    sales = df[df['company'] == company]['sales_value'].dropna().values
    
    # Need minimum 2 values
    if len(sales) >= 2:
        
        # Linear prediction
        x = np.arange(len(sales))
        slope, intercept = np.polyfit(x, sales, 1)
        
        next_year = slope * len(sales) + intercept
        
    else:
        next_year = "Not Enough Data"
    
    forecast_result.append([company, next_year])

# Convert to dataframe
forecast_df = pd.DataFrame(
    forecast_result,
    columns=['company', 'predicted_next_year_revenue']
)

forecast_df.head()
```

```python
# Save forecast results

forecast_df.to_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\output\forecast_revenue.csv",
    index=False
)

print("Forecast saved successfully!")
```
