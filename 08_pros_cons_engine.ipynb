```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\data\clean_analysis_utf8.csv"
)

# Show first rows
df.head()
```

```python
# Create pros and cons

pros_cons = []

for _, row in df.iterrows():
    
    company = row['company']
    
    pro = ""
    con = ""

    # PRO rules
    if row['roe_value'] > 20:
        pro = "Good ROE performance"

    elif row['profit_value'] > 15:
        pro = "Strong profit growth"

    elif row['sales_value'] > 15:
        pro = "Strong sales growth"

    else:
        pro = "Stable performance"

    # CON rules
    if row['sales_value'] < 5:
        con = "Low sales growth"

    elif row['profit_value'] < 5:
        con = "Weak profit growth"

    elif row['stock_value'] < 5:
        con = "Weak stock growth"

    else:
        con = "No major issues"

    pros_cons.append([company, pro, con])

# Convert to dataframe
pros_cons_df = pd.DataFrame(
    pros_cons,
    columns=['company', 'pro', 'con']
)

# Remove duplicates
pros_cons_df = pros_cons_df.drop_duplicates('company')

# Show result
pros_cons_df.head()
```

```python
# Save pros and cons

pros_cons_df.to_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\output\pros_cons.csv",
    index=False
)

print("Pros & Cons saved successfully!")
```
