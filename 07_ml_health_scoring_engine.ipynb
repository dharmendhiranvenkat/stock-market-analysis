```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\data\clean_analysis_utf8.csv"
)

# Show first rows
df.head()
```

```python
# Group company average values

company_data = df.groupby('company')[[
    'sales_value',
    'profit_value',
    'stock_value',
    'roe_value'
]].mean()

# Show table
company_data
```

```python
# Calculate health score

company_data['health_score'] = (
    company_data['sales_value'] * 0.25 +
    company_data['profit_value'] * 0.25 +
    company_data['stock_value'] * 0.20 +
    company_data['roe_value'] * 0.30
)

# Round score
company_data['health_score'] = company_data['health_score'].round(2)

# Show result
company_data[['health_score']]
```

```python
# Create health labels

def health_label(score):
    
    if score >= 85:
        return "EXCELLENT"
    
    elif score >= 70:
        return "GOOD"
    
    elif score >= 50:
        return "AVERAGE"
    
    elif score >= 35:
        return "WEAK"
    
    else:
        return "POOR"

# Apply labels
company_data['health_label'] = company_data['health_score'].apply(health_label)

# Show result
company_data[['health_score', 'health_label']]
```

```python
# Save health scores

company_data.to_csv(
    r"E:\BOLDANAYSTICSCOMPANY\PROJECT2STOCKMARKETANAYSIS\DATASET2\output\health_scores.csv"
)

print("Health scores saved successfully!")
```
